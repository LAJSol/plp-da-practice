```python
# Lab 1.3: Inventory Alerts

inventory = [
  {"name":"Unga", "stock":12, "reorder_level":15},
  {"name":"Sugar", "stock":40, "reorder_level":20},
  {"name":"Milk", "stock":8, "reorder_level":10},
  {"name":"Bread", "stock":25, "reorder_level":10}
]

def restock_report(items):
  """Print stock status and return a list of products to reorder."""
  to_reorder = []
  for item in items:
    if item["stock"] < item["reorder_level"]:
      status ="REORDER"
      to_reorder.append(item["name"])
    else:
      status = "OK"
    print(f"{item['name']:<6}: {item['stock']:>3} {status}")
  return to_reorder

print(restock_report(inventory))
```

#Weekly Sales Report for a Kiosk
**The brief**: Mama Njeri runs a kiosk in Kawangware (a neighbourhood in Nairobi). She records daily sales and customer counts in an exercise book and asks three questions:

1. How much did I make this week?
2. What does a typical customer spend?
3. Which days should I focus on improving?


```python
week = [
    {"day": "Mon", "sales": 1200, "customers": 30},
    {"day": "Tue", "sales": 950,  "customers": 25},
    {"day": "Wed", "sales": 1800, "customers": 41},
    {"day": "Thu", "sales": 1430, "customers": 33},
    {"day": "Fri", "sales": 2100, "customers": 48},
    {"day": "Sat", "sales": 3000, "customers": 62},
    {"day": "Sun", "sales": 2750, "customers": 57},
]
```

```python
def weekly_report(records):
    """Summarise a week of kiosk sales records."""
    total_sales = 0
    total_customers = 0
    for r in records:
        total_sales += r["sales"]
        total_customers += r["customers"]

    average_day = total_sales / len(records)
    avg_basket = total_sales / total_customers

    best_day = records[0]
    weak_days = []
    for r in records:
        if r["sales"] > best_day["sales"]:
            best_day = r
        if r["sales"] < average_day:
            weak_days.append(r["day"])

    print(f"Total sales:        KES {total_sales:,}")
    print(f"Total customers:    {total_customers}")
    print(f"Average per day:    KES {average_day:,.2f}")
    print(f"Avg basket size:    KES {avg_basket:,.2f}")
    print(f"Best day:           {best_day['day']} (KES {best_day['sales']:,})")
    print(f"Below-average days: {', '.join(weak_days)}")

weekly_report(week)
```

#Interpret for  Mama Njeri's kiosk
---

Mama Njeri made KES 13,230 this week. Each customer spends about KES 45 on average. Weekends are your strongest days, and Monday to Thursday are all below average. Since the basket size is small, the biggest opportunity is to raise weekday traffic, for example with a weekday bundle offer on bread and milk.
